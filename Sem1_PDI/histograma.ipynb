```python
import requests
from PIL import Image
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from io import BytesIO
```

```python
image_url = "https://www.buenoyvegano.com/wp-content/uploads/2020/01/Gato-apropiado.jpg"  # Reemplaza con una URL válida
```

```python
def download_image(url):
    response = requests.get(url)
    response.raise_for_status()
    return np.array(Image.open(BytesIO(response.content)))

def adjust_brightness(image, brightness):
    return np.clip(image * brightness, 0, 255).astype(np.uint8)

def adjust_contrast(image, contrast):
    mean = np.mean(image)
    return np.clip((image - mean) * contrast + mean, 0, 255).astype(np.uint8)


def display_image_and_histogram(image, brightness=1.0, contrast=1.0):
    adjusted_image = adjust_brightness(image, brightness)
    adjusted_image = adjust_contrast(adjusted_image, contrast)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    axes[0].imshow(adjusted_image)
    axes[0].axis('off')
    axes[0].set_title("Imagen Ajustada")

    grayscale_image = np.mean(adjusted_image, axis=2).astype(np.uint8) if adjusted_image.ndim == 3 else adjusted_image
    sns.histplot(grayscale_image.flatten(), bins=256, kde=False, ax=axes[1], color="blue")
    axes[1].set_title("Histograma")
    axes[1].set_xlabel("Nivel de Intensidad")
    axes[1].set_ylabel("Frecuencia")

    plt.tight_layout()
    plt.show()

```

```python
image = download_image(image_url)
interact(
    lambda brightness, contrast: display_image_and_histogram(image, brightness, contrast),
    brightness=FloatSlider(value=1.0, min=0.5, max=2.0, step=0.1, description="Brillo"),
    contrast=FloatSlider(value=1.0, min=0.5, max=2.0, step=0.1, description="Contraste")
)
```

```python
def display_image_and_normalized_histogram(image, brightness=1.0, contrast=1.0):
    # Ajustar brillo y contraste manualmente
    adjusted_image = adjust_brightness(image, brightness)
    adjusted_image = adjust_contrast(adjusted_image, contrast)

    # Mostrar la imagen ajustada
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    axes[0].imshow(adjusted_image)
    axes[0].axis('off')
    axes[0].set_title("Imagen Ajustada")

    # Mostrar el histograma normalizado
    grayscale_image = np.mean(adjusted_image, axis=2).astype(np.uint8) if adjusted_image.ndim == 3 else adjusted_image
    pixel_values = grayscale_image.flatten()
    sns.histplot(pixel_values, bins=256, kde=False, ax=axes[1], color="blue", stat="probability")
    axes[1].set_title("Histograma Normalizado")
    axes[1].set_xlabel("Nivel de Intensidad")
    axes[1].set_ylabel("Frecuencia Relativa")

    plt.tight_layout()
    plt.show()
```

```python
interact(
    lambda brightness, contrast: display_image_and_normalized_histogram(image, brightness, contrast),
    brightness=FloatSlider(value=1.0, min=0.5, max=2.0, step=0.1, description="Brillo"),
    contrast=FloatSlider(value=1.0, min=0.5, max=2.0, step=0.1, description="Contraste")
)
```
