## Binarización de Otzu
[Opencv code](https://docs.opencv.org/4.x/d7/d4d/tutorial_py_thresholding.html)

[Explicación](https://hbyacademic.medium.com/mathematical-secrets-behind-image-binarization-otsu-thresholding-25edf8d7cb60)

* +1 punto explicar la siguiente clase la matemática de este modelo (opcional).

```python
def floyd_steinberg_dithering(image):
    """Aplica el dithering de Floyd-Steinberg a una imagen en escala de grises y la binariza"""
    img = image.astype(np.float32)  # Convertir a flotante para manipular errores

    for y in range(img.shape[0] - 1):
        for x in range(1, img.shape[1] - 1):
            old_pixel = img[y, x]
            new_pixel = 255 if old_pixel > 127 else 0  # Umbralización binaria
            img[y, x] = new_pixel
            error = old_pixel - new_pixel  # Calcular error de cuantización

            # Distribuir el error a los píxeles vecinos
            img[y, x + 1] += error * 7 / 16
            img[y + 1, x - 1] += error * 3 / 16
            img[y + 1, x] += error * 5 / 16
            img[y + 1, x + 1] += error * 1 / 16

    return np.clip(img, 0, 255).astype(np.uint8)  # Asegurar valores válidos
```

```python
# Aplicar Floyd-Steinberg dithering
dithered_image = floyd_steinberg_dithering(gray_image)
```

```python
# Mostrar resultados
plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.title("Imagen Original")
plt.imshow(i1, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Escala de Grises")
plt.imshow(gray_image, cmap="gray")
plt.axis("off")

plt.subplot(1, 3, 3)
#plt.title("Dithering Binario")
plt.imshow(dithered_image, cmap="gray")
plt.axis("off")

plt.show()
```
